```python
import json
import plotly.graph_objects as go
import pandas as pd
```

```python
with open("life.json", "r", encoding="utf-8") as f:
    life = json.load(f)
```

```python
with open("locations.json", "r", encoding="utf-8") as f:
    locations = json.load(f)
```

```python

# build sorted numeric arrays
life_sorted = sorted(life, key=lambda d: int(d["year"]))
year  = [int(d["year"]) for d in life_sorted]
work   = [float(d["work"]) for d in life_sorted]
money  = [float(d["money"]) for d in life_sorted]
family = [float(d["family"]) for d in life_sorted]
physical_work = [float(d["physical_work"]) for d in life_sorted]
traveling = [float(d["traveling"]) for d in life_sorted]
gym = [float(d["gym"]) for d in life_sorted]
windows = [float(d["windows"]) for d in life_sorted]
linux = [float(d["linux"]) for d in life_sorted]
python = [float(d["python"]) for d in life_sorted]
#location_id = [float(d["location_id"]) for d in life_sorted]
#event = [float(d["event"]) for d in life_sorted]

fig = go.Figure()
fig.add_trace(go.Scatter(
    x=year,
    y=work,
    mode="lines+markers",
    line=dict(width=7),  # thinner/normal
    marker=dict(size=4),
    name="Work"
    ))
fig.add_trace(go.Scatter(
    x=year,
    y=money,
    mode="lines+markers",
    line=dict(width=4),  # thinner/normal
    marker=dict(size=8),
    name="Money"
    ))


fig.add_trace(go.Scatter(
    x = year,
    y = physical_work,
    mode = "lines+markers",
    name = "Phisical Work"
))



fig.update_layout(
    yaxis=dict(type="linear", range=[0, 11], autorange=False, dtick=1),
    xaxis=dict(tickmode="linear")
)

fig.show()

```

```python
fig.write_html("work_life.html", include_plotlyjs="cdn")
```
