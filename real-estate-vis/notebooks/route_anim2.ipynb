```python
# importing libary's

import folium
import osmnx as ox
```

```python

# getting lat and lon of the location and generating a map
m = folium.Map(
    location=[14.442055, 120.998021], # getting the location
    zoom_start=14 # zoom on the map
)

m
```

```python
# Locations 
Las_Piñas = [14.442055, 120.998021]
Tagaytay = [14.09920, 120.94025]
Calatagan = [13.82181, 120.62812]
```

```python
locations = [
    Las_Piñas,
    Tagaytay,
    Calatagan
]

lats = [location[0] for location in locations]
lons = [location[1] for location in locations]

north = max(lats) + 0.1
south = min(lats) - 0.1
east = max(lons) + 0.1
west = min(lons) - 0.1

print()
print(west)
print(east)
print(south)
print(north)
print()
```

```python
G = ox.graph.graph_from_bbox(
    (west, south, east, north),
    network_type="drive"
)

print("Nodes:", len(G.nodes))
print("Road:", len(G.edges))
```

```python
# getting nearest point to the location for the road
lp_node = ox.distance.nearest_nodes(
    G,
    X=Las_Piñas[1],
    Y=Las_Piñas[0]
)

t_node = ox.distance.nearest_nodes(
    G,
    X=Tagaytay[1],
    Y=Tagaytay[0]
)

c_node = ox.distance.nearest_nodes(
    G,
    X=Calatagan[1],
    Y=Calatagan[0]
)
```

```python
print("Las Piñas:", lp_node)
print("Tagaytay:", t_node)
print("Calatagan:", c_node)
```

```python
# getting the path to the locations
route_lp_tagaytay = ox.routing.shortest_path(
    G,
    lp_node,
    t_node,
    weight="length"
)

route_tagaytay_cal = ox.routing.shortest_path(
    G,
    t_node,
    c_node,
    weight="length"
)

print("Las_Piñas - Tagaytay:", len(route_lp_tagaytay), "road")
print("Tagaytay - Calatagan:", len(route_tagaytay_cal), "road")
```

```python
# converting osmnx node into lat lon

route1_coords = [
    (G.nodes[node]["y"], G.nodes[node]["x"])
    for node in route_lp_tagaytay
]

route2_coords = [
    (G.nodes[node]["y"], G.nodes[node]["x"])
    for node in route_tagaytay_cal
]

# printing it onto screen
route1_coords
route2_coords
```

```python
folium.PolyLine(
    locations=route1_coords,
    color="blue",
    weight=5,
    opacity=0.5,
    tooltip="Las_Piñas - Tagaytay"
).add_to(m) # adding it to the map

folium.PolyLine(
    locations=route2_coords,
    color="blue",
    weight=5,
    opacity=0.5,
    tooltip="Tagaytay - Calatagan"
).add_to(m)

m
```
