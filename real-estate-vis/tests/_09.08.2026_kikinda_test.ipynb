```python
import osmnx as ox
import matplotlib.pyplot as plt
```

```python
city = "City of Kikinda, Serbia"
```

```python
G = ox.geocode_to_gdf(city)
```

```python
print(G)
```

```python
G.plot()
```

```python
roads = ox.graph.graph_from_place(
    city,
    network_type="all"
)
```

```python
nodes, edges = ox.convert.graph_to_gdfs(roads)
```

```python
nodes.plot(
    figsize=(10,10),
    facecolor="lightblue",
    edgecolor="black"
)
```

```python
fig, ax = ox.plot_graph(
    roads,
    node_size=0,
    edge_color="black",
    edge_linewidth=0.8,
    bgcolor="white",
    show=False,
    close=False
)
```

```python
fig, ax = plt.subplots(figsize=(12,12))
```

```python
fig, ax = ox.plot_graph(
    roads,
    node_size=0,
    edge_color="#76895E",
    edge_linewidth=0.8,
    bgcolor="none",
    show=False,
    close=False
)

G.plot(
    ax=ax,
    facecolor="none",      
    edgecolor="#9B4743",
    linewidth=2,
    zorder=5
)

ax.set_axis_off()
plt.show()
```

```python
tags = {
    "amenity": [
        "university",
        "hospital",
        "school",
        "library",
        "museum",
        "theatre",
        "attraction",
        "gallery"
    ]
}

places = ox.features.features_from_place(
    city,
    tags=tags
)

places.head()
```

```python
names = places["name"]

for n in places["amenity"]:
    print(n)

for n in names:
    print(n)
```

```python
fig, ax = ox.plot_graph(
    roads,
    node_size=0,
    edge_color="#76895E",
    edge_linewidth=0.8,
    bgcolor="none",
    show=False,
    close=False
)

G.plot(
    ax=ax,
    facecolor="none",      
    edgecolor="#9B4743",
    linewidth=2,
    zorder=5
)

categories = {
    "museum": "purple",
    "library": "blue",
    "hospital": "red",
    "school": "orange",
    "theatre": "green"
}


for amenity, color in categories.items():

    n = places[places["amenity"] == amenity]

    if not n.empty:
        n.plot(
            ax=ax,
            color=color,
            markersize=50,
            label=amenity.capitalize(),
            zorder=10
        )

ax.legend(
    title="Amenities",
    loc="upper right",
    frameon=True
)


ax.set_axis_off()
plt.show()


print(places[["name", "amenity"]].value_counts())
```
