```python
import folium
import numpy as np
import osmnx as ox
import matplotlib.pyplot as plt
```


custom_filter = (
    '["highway"~"motorway|trunk|primary|secondary|tertiary"]'
)

G = ox.graph_from_point(
    (14.442055, 120.998021),
    dist=25000,
    network_type="drive",
    custom_filter=custom_filter,
    simplify=True
)

print("Nodes:", len(G.nodes))
print("Edges:", len(G.edges))

```python

fig, ax = ox.plot_graph(
    G,
    figsize=(12, 12),
    node_size=0,
    edge_color="steelblue",
    edge_linewidth=0.8,
    bgcolor="white",
    show=False,
    close=False
)

ax.scatter(
    120.998021,
    14.442055,
    s=150,
    color="red",
    edgecolor="white",
    linewidth=2,
    zorder=10
)

ax.set_title("Road Network — 5 km around Las Piñas")

plt.show()
```
