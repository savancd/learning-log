```python
import folium
```

```python
TILES = {
    "dark": {
        "url": "OpenStreetMap",
        "attr": "OpenStreetMap contributors"
    },
    "satellite": {
        "url": (
            "https://server.arcgisonline.com/ArcGIS/rest/"
            "services/World_Imagery/MapServer/tile/{z}/{y}/{x}"
        ),
        "attr": "Esri"
    },
}

```

```python

def create_map(center, zoom=13, style="dark", interactive=True):
    tile = TILES[style]

    tile_url = tile["url"]

    # Folium recognizes "OpenStreetMap" as a built-in provider.
    if tile_url == "OpenStreetMap":
        m = folium.Map(
            location=center,
            zoom_start=zoom,
            tiles="OpenStreetMap",
            control_scale=interactive,
        )

    else:
        # Custom URL, such as the Esri satellite layer.
        m = folium.Map(
            location=center,
            zoom_start=zoom,
            tiles=None,
            control_scale=interactive,
        )

        folium.TileLayer(
            tiles=tile_url,
            attr=tile["attr"],
            name=style.title(),
            overlay=False,
            control=interactive,
        ).add_to(m)

    return m
```

```python
center = [14.444979, 120.999261]
```

```python
m = create_map(
    center=center,
    zoom=11,
    style="dark",
    interactive=False,
)

m
```
