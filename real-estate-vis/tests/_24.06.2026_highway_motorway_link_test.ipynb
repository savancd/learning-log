```python

features = data["features"]

# collect all coords from all features (LineString case)
all_latlon = []
for f in features:
    geom = f["geometry"]
    for lon, lat in geom["coordinates"]:
        all_latlon.append((lat, lon))

center = all_latlon[len(all_latlon)//2]

m = folium.Map(location=center, zoom_start=11, tiles="OpenStreetMap")

for f in features:
    geom = f["geometry"]
    latlon = [(lat, lon) for lon, lat in geom["coordinates"]]
    folium.PolyLine(latlon, color="blue", weight=3, opacity=0.8).add_to(m)

m
```
