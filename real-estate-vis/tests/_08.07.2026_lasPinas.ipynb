## Getting the location from a .json file

```python
# importing the json to read a file
import json
```

```python
# getting a file from path
location_path = Path("/data/repo-learning/computer-Science-and-Software-Engineering/real-estate-vis/datasets/locations/las_pinas.json")
```

```python
# reading the file
# this is converting into dict
location = json.loads(location_path.read_text())
```

```python
# printing the location 
location
```

```python
# can print just the keys of the file
location.keys()
```

```python
# getting just the name of location
location["city"]
```

<strong>All the files are previously loaded into memory without any visualization yet</strong>

```python
location.items()
```

```python
for key, value in location.items():

    placeholder = "{{" + key + "}}"

    query = query.replace(
        placeholder,
        str(value)
    )

#    print(placeholder)
```

```python
import requests
```

```python
# sending a requests
response = requests.post(
    "https://overpass-api.de/api/interpreter",
    data=query
)
```
